```python
# Handling missing values in TotalCharges
df['TotalCharges']=df['TotalCharges'].str.strip()
df['TotalCharges']=df['TotalCharges'].replace("",np.nan)
print('Missing Values after handling the blank rows:')
print(df['TotalCharges'].isnull().sum())
```

```python
# Converting the TotalCharges column into numeric
df['TotalCharges']=pd.to_numeric(df['TotalCharges'],errors='coerce')
print(df['TotalCharges'].dtype)
```

```python
df[df['TotalCharges'].isnull()][[
    "customerID",
    "tenure",
    "MonthlyCharges",
    "TotalCharges",
    "Churn"
]]
```

## Handling Missing TotalCharges

The `TotalCharges` column initially contained 11 blank values and was stored as an object data type.

Investigation showed that all 11 affected customers had a tenure of 0 months, while valid MonthlyCharges values were present. Since TotalCharges represents cumulative charges and these customers had zero months of tenure, the missing values were treated as 0.

The affected customer records were retained rather than removed to avoid unnecessary loss of valid observations.

```python
# Handling the missing values for zero tentures
zero_tentures=(df['tenure']==0 & df['TotalCharges'].isnull())
print('Total Zero_tentures:',zero_tentures.sum())
df.loc[zero_tentures,'TotalCharges']=0
```

# Exploratory Data Analysis

## Business Questions

The exploratory analysis is structured around business questions related to customer churn.

The objective is to identify patterns and customer segments associated with higher observed churn and translate those findings into actionable retention opportunities.

The analysis focuses on association and observed patterns rather than causal claims.


```python
# Question 1: What is the overall churn situation?
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(5,5))
sns.countplot(
    data=df,
    x='Churn',
)
plt.xlabel('Churn')
plt.ylabel('Count of Churned Customers')
plt.title('Churned Customer Distribution')
plt.show()
```

```python
churn_summary=(df['Churn'].value_counts().rename_axis("Churn").reset_index(name='Customers'))
churn_summary['Percentage']=((churn_summary['Customers']/len(df))*100).round(2)
churn_summary
```

### Insight

The dataset contains 7,043 customers, of which 1,869 customers (26.54%) are classified as churned, while 5,174 customers (73.46%) are classified as retained.

This indicates that approximately one-quarter of the customers in the dataset have churned. This provides the baseline churn level against which differences across customer characteristics and service segments can be compared.

This analysis describes an observed association in the dataset and does not establish the causes of customer churn.

```python
# Question 2: Does customer tenure relate to churn?
```

```python
df.head()
```

```python
tenture_summary=(df.groupby('Churn')['tenure'].agg(['count','min','max','mean','median']).round(2))
tenture_summary
```

```python
#box plot
plt.figure(figsize=(5,5))
sns.boxplot(
    data=df,
    x='Churn',
    y='tenure'
)
plt.title('Customer Tenure by Churn Status')
plt.xlabel('Churn Status')
plt.ylabel('Tenure')

plt.show()
```

```python
df['tenure_group']=pd.cut(
    df['tenure'],
    bins=[-1,12,24,48,72],
    labels=[
        '0-12 months',
        '13-24 months',
        '25-48 months',
        '49-72 months'
    ]
)
```

```python
tenure_churn=(pd.crosstab(df['tenure_group'],df['Churn'],normalize='index').mul(100).round(2))
tenure_churn
```

### Insight

Customer tenure shows a clear association with observed churn in the dataset.

Customers with shorter tenure have substantially higher observed churn rates. The 0–12 month tenure group has an observed churn rate of 47.44%, compared with 28.71% for customers with 13–24 months of tenure, 20.39% for customers with 25–48 months, and only 9.51% for customers with 49–72 months.

The average tenure of churned customers is also considerably lower at 17.98 months compared with 37.57 months among retained customers, while the median tenure is 10 months for churned customers versus 38 months for retained customers.

This suggests that early-tenure customers represent an important segment for retention analysis. However, the analysis shows association rather than causation and does not establish that low tenure itself causes churn.

```python
#Does contract type relate to customer churn?
contract_churn=(
    pd.crosstab(df['Contract'],df['Churn'],normalize='index').mul(100).round(2)
)
contract_churn
```

```python
contract_summary=(
    df.groupby('Contract')['Churn']
    .apply(lambda x: (x=='Yes').mean()*100).round(2)
    .reset_index(name='Churn_rate')
    .sort_values('Churn_rate',ascending=False)
)
contract_summary
```

```python
plt.figure(figsize=(8,5))
sns.barplot(
    data=contract_summary,
    x='Contract',
    y='Churn_rate'
)
plt.xlabel('Contract Type')
plt.ylabel('Churn Rate')
plt.title('Churn Rate by Contract Type')
plt.show()
```

### Insight

Contract type shows a strong association with observed customer churn.

Customers on month-to-month contracts have an observed churn rate of 42.71%, compared with 11.27% for one-year contracts and 2.83% for two-year contracts.

The substantially higher churn rate among month-to-month customers identifies this group as an important segment for further investigation and potential retention analysis.

However, contract type may be related to other customer characteristics such as tenure, pricing, services or customer profile. Therefore, this analysis should be interpreted as an observed association rather than evidence that contract type itself causes churn.

```python
# Does payment method show differences in observed churn?
payment_churn=pd.crosstab(
    df['PaymentMethod'],
    df['Churn'],
    normalize='index'
).mul(100).round(2)
payment_churn
```

```python
payment_summary=(
    df.groupby('PaymentMethod')['Churn'].apply(lambda x: (x=='Yes').mean()*100).
    round(2)
    .reset_index(name='Churn Rate')
    .sort_values('Churn Rate',ascending=False)
)
payment_summary
```

```python
plt.figure(figsize=(8,5))
sns.barplot(
    data=payment_summary,
    x='PaymentMethod',
    y='Churn Rate'
    )
plt.xlabel('Payment Method')
plt.ylabel('Churn Rate')
plt.title('Churn Rate by Payment Method')
plt.show()
```

### Insight

Payment method shows a substantial association with observed customer churn.

Customers using electronic check have an observed churn rate of 45.29%, which is considerably higher than customers using mailed check (19.11%), automatic bank transfer (16.71%), or automatic credit card payment (15.24%).

The electronic-check segment therefore represents an important customer group for further investigation and potential retention analysis. However, payment method alone should not be interpreted as a causal driver of churn because it may be associated with other customer characteristics such as contract type, tenure and service usage.

```python
# now creating a segment summary
segment_summary=(
    df.groupby(['tenure_group','Contract']).
    agg(
        customers=('customerID','count'),
        churn=('Churn',lambda x: (x=="Yes").sum()),
        churned_rate=('Churn',lambda x: (x=='Yes').mean()*100)
    ).reset_index()
)
segment_summary['churned_rate']=segment_summary['churned_rate'].round(2)
segment_summary=segment_summary.sort_values('churned_rate',ascending=False)
segment_summary
```

```python
plt.figure(figsize=(8,5))
sns.barplot(
    data=segment_summary,
    x='tenure_group',
    y='churned_rate',
    hue='Contract'
)
plt.xlabel('Tenure Group')
plt.ylabel('Churn Rate')
plt.title('Churned Rate by Tenure Group and Contract Type')
plt.legend(title='Contract Type')
plt.tight_layout()
plt.show()
```


###Customer Segmentation Insight:
Customer churn varies substantially across tenure and contract segments. The highest observed churn rate is among 0–12 month customers on month-to-month contracts (51.35%), followed by month-to-month customers in the 13–24 month (37.72%) and 25–48 month (32.92%) tenure groups. In comparison, two-year contract segments show very low observed churn, ranging from 0% to 3.33%. The results indicate that early-tenure, month-to-month customers represent an important segment for further retention analysis. However, these results describe associations in the observed dataset and do not establish that contract type or tenure directly causes churn.

```python
# now identifying high risk segments
high_risk_segments=(segment_summary[segment_summary['customers']>=100].sort_values('churned_rate',ascending=False))
high_risk_segments
```

```python
high_risk=df[
    (df['tenure_group']=='0-12 months')&
    (df['Contract']=='Month-to-month')&
    (df['PaymentMethod']=='Electronic check')
]
high_risk_summary={
    'customers':len(high_risk),
    'churned':(high_risk['Churn']=='Yes').sum(),
    'churn_rate':round((high_risk['Churn']=='Yes').mean()*100,2)
}
high_risk_summary
```

###Combined Segment Insight:
Customers in the 0–12 month tenure group with month-to-month contracts and electronic-check payment have an observed churn rate of 63.10%, with 602 of 954 customers having churned. This is substantially higher than the overall observed churn rate of 26.54%, indicating that this combined segment warrants focused retention analysis. However, the result represents an association in the observed dataset and does not establish that these characteristics independently cause churn.

## Limitations

- The analysis identifies associations between customer characteristics and observed churn but does not establish causal relationships.
- The dataset represents a specific customer population and may not generalize to other companies, markets, or time periods.
- The analysis is primarily descriptive and does not include predictive churn modeling.
- Customer satisfaction, service interaction history, competitor activity, and other potentially important churn factors are not available in the dataset.
- The analysis does not measure the effectiveness or financial impact of the proposed retention actions.
- Some customer segments may have smaller sample sizes, so their observed churn rates should be interpreted cautiously.

### Next Steps

- Collect additional customer-level information such as support interactions, satisfaction scores, complaints, and service issues.
- Investigate the reasons behind the high churn observed among early-tenure and month-to-month customers.
- Evaluate retention strategies through controlled experiments or A/B testing.
- Develop a churn prediction model to identify individual customers who may be at higher risk of leaving.
- Estimate the potential financial impact of different retention strategies.
- Monitor churn rates over time to determine whether retention initiatives produce measurable improvements.

## Key Decisions

- Missing `TotalCharges` values were investigated and found only for customers with zero tenure. These values were set to 0 after reviewing the observed data pattern.
- `tenure_group` was created to support interpretable customer segmentation.
- Customer segments were analyzed using tenure and contract type.
- A combined high-risk segment was identified using tenure, contract type, and payment method.
- Findings are treated as associations rather than causal relationships.
